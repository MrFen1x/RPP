## Задание по варианту
Разработать программу, которая удаляет цепочки четных чисел из списка `A`, если хотя бы одно из чисел в цепочке содержится в списке `B`. Программа должна предусматривать ввод данных с клавиатуры и их автоматическую генерацию, а также корректно обрабатывать ошибки ввода.

## Описание реализации
Программа реализована на языке Python и состоит из следующих функций:

1. **`input_list(prompt)`** – ввод списка чисел с клавиатуры с обработкой ошибок.
2. **`generate_list(size, min_val, max_val)`** – генерация случайного списка заданного размера в указанном диапазоне.
3. **`input_a()`** – выбор способа ввода списка `A` (вручную или генерация).
4. **`input_b()`** – выбор способа ввода списка `B`.
5. **`remove_even_chains(a, b)`** – удаление цепочек четных чисел из `A`, если в них есть хотя бы одно число из `B`.
6. **`main()`** – основная логика программы, управление вводом и выводом данных.

Алгоритм работы:
1. Пользователь выбирает способ ввода для списка `A` (вручную или генерация).
2. Пользователь вводит список `B` аналогично `A`.
3. Программа анализирует список `A`, находя последовательности четных чисел.
4. Если в последовательности есть число из `B`, она удаляется.
5. Выводится итоговый список `A` после обработки.

## Описание возникших затруднений
В процессе разработки возникли следующие затруднения:
- Обработка ввода данных: требовалось корректно распознавать и предотвращать ошибки при вводе чисел пользователем.
- Удаление цепочек четных чисел: изначально программа просто удаляла все четные числа, но позже была доработана для работы с последовательностями.
- Проверка наличия элементов из `B` в четных цепочках: использовалось множество `set(B)` для оптимизации поиска.

## Описание альтернативных способов решения
Возможны другие подходы к решению задачи:
1. **Использование рекурсии** – можно было бы использовать рекурсивный алгоритм для нахождения и удаления цепочек.
2. **Работа с индексами** – вместо формирования временных списков можно было бы изменять `A` по индексам.
3. **Использование библиотеки NumPy** – для быстродействия можно было бы использовать векторизированные операции.

## Вывод
В ходе выполнения лабораторной работы была разработана программа на Python, корректно удаляющая цепочки четных чисел из списка `A`, если они содержат элементы списка `B`. Программа поддерживает обработку ошибок ввода и позволяет использовать как ручной, так и автоматический ввод данных.



````python
import random  # Импорт модуля для работы со случайными числами


def input_list(prompt):
    """Функция для ввода списка чисел с клавиатуры."""
    while True:
        try:
            return list(map(int, input(prompt).split()))  # Преобразуем ввод в список чисел
        except ValueError:
            print("Ошибка ввода! Введите числа через пробел.")  # Обработка ошибки, если введены нечисловые данные


def generate_list(size, min_val, max_val):
    """Функция для генерации списка случайных чисел."""
    return [random.randint(min_val, max_val) for _ in
            range(size)]  # Создаём список случайных чисел в заданном диапазоне


def input_a():
    """Функция для выбора способа ввода списка A."""
    choice = input("Выберите способ ввода списка A (1 - вручную, 2 - генерация): ")

    if choice == '1':
        a = input_list("Введите список A: ")  # Ввод списка вручную
    elif choice == '2':
        size = int(input("Введите размер списка A: "))  # Запрашиваем размер списка
        min_val, max_val = map(int, input("Введите диапазон значений (min max): ").split())  # Вводим диапазон значений
        a = generate_list(size, min_val, max_val)  # Генерируем случайный список
        print("Сгенерированный список A:", a)
    else:
        print("Ошибка выбора. Завершение программы.")  # Если введён неправильный выбор — завершаем программу
        return None

    return a


def input_b():
    """Функция для выбора способа ввода списка B."""
    choice = input("Выберите способ ввода списка B (1 - вручную, 2 - генерация): ")

    if choice == '1':
        b = input_list("Введите список B: ")  # Ввод списка вручную
    elif choice == '2':
        size = int(input("Введите размер списка B: "))  # Запрашиваем размер списка
        min_val, max_val = map(int, input("Введите диапазон значений (min max): ").split())  # Вводим диапазон значений
        b = generate_list(size, min_val, max_val)  # Генерируем случайный список
        print("Сгенерированный список B:", b)
    else:
        print("Ошибка выбора. Завершение программы.")  # Если введён неправильный выбор — завершаем программу
        return None

    return b


def remove_even_chains(a, b):
    """Функция для удаления цепочек четных чисел, если в них есть хотя бы один элемент из B."""
    b_set = set(b)  # Преобразуем B в множество для быстрого поиска
    result = []
    temp_chain = []

    for num in a:
        if num % 2 == 0:  #Ищем четную цепочку в списке А
            temp_chain.append(num)
        else:
            if temp_chain:
                if not any(x in b_set for x in temp_chain):  # Если в списке В нет элементов из цепочки
                    result.extend(temp_chain)  # Добавляем цепочку в результат
                temp_chain = []  #Если такие элементы есть очищаем цепочку
            result.append(num)  #Добавляем нечетное число в результат

    if temp_chain and not any(x in b_set for x in temp_chain):
        result.extend(temp_chain)

    return result


def main():
    """Главная функция программы."""
    a = input_a()  # Ввод списка A
    if a is None:  # Проверяем, не прервали ли ввод
        return

    b = input_b()  # Ввод списка B
    if b is None:  # Проверяем, не прервали ли ввод
        return

    result = remove_even_chains(a, b)  # Удаляем цепочки четных чисел, содержащие элементы из B

    print("Результат:", result)  # Выводим итоговый список A


if __name__ == "__main__":
    main()  # Запуск программы
````

## Описание реализации

Она состоит из следующих функций:

## Описание возникших затруднений

## Вывод
В ходе выполнения лабораторной работы была реализована программа, выполняющая генерацию матрицы, вычисление сумм элементов и определение долей сумм по столбцам. Программа корректно сохраняет результаты в файл и использует библиотеку NumPy для оптимизации вычислений.


````python
import numpy as np  # Импорт библиотеки NumPy для работы с массивами


def generate_matrix(n, m, min_val=0, max_val=100):
    """Функция для генерации матрицы размера N x M со случайными значениями."""
    return np.random.randint(min_val, max_val + 1, size=(n, m))  # Создание матрицы случайных целых чисел


def save_results(matrix, total_sum, column_sums, result_matrix, filename="matrix_results.txt"):
    """Сохраняет исходную матрицу и результаты вычислений в файл."""
    with open(filename, "w", encoding="utf-8") as file:
        file.write("Сгенерированная матрица:\n")  # Запись заголовка в файл
        np.savetxt(file, matrix, fmt="%d")  # Сохранение матрицы в файл в формате целых чисел
        file.write(f"\nСумма всех элементов матрицы: {total_sum}\n")  # Запись суммы всех элементов
        file.write("\nМатрица с долями суммы по столбцам:\n")  # Заголовок для итоговой матрицы
        np.savetxt(file, result_matrix, fmt="%.4f")  # Сохранение итоговой матрицы с долями


def main():
    """Основная функция программы."""
    n = int(input("Введите количество строк (N): "))  # Ввод количества строк
    m = int(input("Введите количество столбцов (M): "))  # Ввод количества столбцов

    matrix = generate_matrix(n, m)  # Генерация случайной матрицы
    print("Сгенерированная матрица:")
    for x in matrix: print(x)

    total_sum = np.sum(matrix)  # Вычисление суммы всех элементов матрицы
    print(f"\nСумма всех элементов матрицы: {total_sum}\n")  # Вывод суммы

    column_sums = np.sum(matrix, axis=0)  # Вычисление суммы элементов каждого столбца
    result_matrix = [column_sums / total_sum]  # Добавление строки с долями сумм столбцов

    print("Матрица с долями суммы по столбцам:")
    for x in result_matrix: print(x)

    save_results(matrix, total_sum, column_sums, result_matrix)  # Сохранение результатов в файл
    print("\nРезультаты сохранены в файл matrix_results.txt")  # Сообщение о сохранении


if __name__ == "__main__":
    main()  # Запуск основной функции
````

````python
import os
import csv
from datetime import datetime

# 1. Подсчёт количества файлов в директории
def count_files_in_directory(directory_path):
    try:
        if not os.path.isdir(directory_path):
            raise FileNotFoundError(f"Директория {directory_path} не существует.")

        file_count = sum(1 for entry in os.scandir(directory_path) if entry.is_file())
        print(f"Количество файлов в директории '{directory_path}': {file_count}")
        return file_count
    except Exception as e:
        print(f"Ошибка при подсчёте файлов: {e}")
        return 0

# 2. Работа с файлом data.csv
# 2.0. Чтение данных из файла в список словарей
def read_trips_from_csv(file_path):
    trips = []
    try:
        with open(file_path, mode='r', encoding='utf-8') as file:
            reader = csv.DictReader(file)
            for row in reader:
                row['№'] = int(row['№'])  # Преобразуем № в int
                # Преобразуем строку даты в объект datetime для сортировки и фильтрации
                row['дата и время'] = datetime.strptime(row['дата и время'], '%Y-%m-%d %H:%M:%S')
                trips.append(row)
        return trips
    except FileNotFoundError:
        print(f"Файл {file_path} не найден.")
        return []
    except Exception as e:
        print(f"Ошибка при чтении файла: {e}")
        return []

# 2.1. Вывод с сортировкой по строковому полю (марка автомобиля)
def print_trips_sorted_by_brand(trips):
    if not trips:
        print("Нет данных для сортировки.")
        return
    sorted_trips = sorted(trips, key=lambda x: x['марка автомобиля'])
    print("\nПроезды, отсортированные по марке автомобиля:")
    for trip in sorted_trips:
        print(f"№: {trip['№']}, Дата и время: {trip['дата и время'].strftime('%Y-%m-%d %H:%M:%S')}, "
              f"Номерной знак: {trip['номерной знак']}, Марка: {trip['марка автомобиля']}")

# 2.2. Вывод с сортировкой по числовому полю (№)
def print_trips_sorted_by_number(trips):
    if not trips:
        print("Нет данных для сортировки.")
        return
    sorted_trips = sorted(trips, key=lambda x: x['№'])
    print("\nПроезды, отсортированные по номеру:")
    for trip in sorted_trips:
        print(f"№: {trip['№']}, Дата и время: {trip['дата и время'].strftime('%Y-%m-%d %H:%M:%S')}, "
              f"Номерной знак: {trip['номерной знак']}, Марка: {trip['марка автомобиля']}")

# 2.3. Вывод с фильтрацией по критерию (дата после 2023-10-15)
def print_trips_filtered_by_date(trips):
    if not trips:
        print("Нет данных для фильтрации.")
        return
    filter_date = datetime.strptime("2023-10-15 00:00:00", '%Y-%m-%d %H:%M:%S')
    filtered_trips = [trip for trip in trips if trip['дата и время'] > filter_date]
    print("\nПроезды после 2023-10-15:")
    for trip in filtered_trips:
        print(f"№: {trip['№']}, Дата и время: {trip['дата и время'].strftime('%Y-%m-%d %H:%M:%S')}, "
              f"Номерной знак: {trip['номерной знак']}, Марка: {trip['марка автомобиля']}")

# 3. Добавление нового проезда и сохранение в файл
def add_trip_and_save(file_path, trips):
    try:
        # Получаем новый номер (максимальный + 1)
        new_number = max(trip['№'] for trip in trips) + 1 if trips else 1

        # Запрашиваем данные у пользователя
        print("\nДобавление нового проезда:")
        date_time = input("Введите дату и время (ГГГГ-ММ-ДД ЧЧ:ММ:СС): ")
        license_plate = input("Введите номерной знак: ")
        car_brand = input("Введите марку автомобиля: ")

        # Преобразуем дату в datetime
        date_time_obj = datetime.strptime(date_time, '%Y-%m-%d %H:%M:%S')

        # Добавляем новый проезд
        new_trip = {
            '№': new_number,
            'дата и время': date_time_obj,
            'номерной знак': license_plate,
            'марка автомобиля': car_brand
        }
        trips.append(new_trip)

        # Сохраняем все данные обратно в файл
        with open(file_path, mode='w', encoding='utf-8', newline='') as file:
            fieldnames = ['№', 'дата и время', 'номерной знак', 'марка автомобиля']
            writer = csv.DictWriter(file, fieldnames=fieldnames)
            writer.writeheader()
            # Преобразуем datetime обратно в строку при записи
            for trip in trips:
                trip_copy = trip.copy()
                trip_copy['дата и время'] = trip['дата и время'].strftime('%Y-%m-%d %H:%M:%S')
                writer.writerow(trip_copy)
        print("Новый проезд успешно добавлен и сохранён в файл.")
    except Exception as e:
        print(f"Ошибка при сохранении данных: {e}")

# Основная программа
def main():
    # 1. Подсчёт файлов в директории
    directory_path = "."  # Текущая директория (можно изменить)
    count_files_in_directory(directory_path)

    # 2. Работа с файлом data.csv
    file_path = "data.csv"
    trips = read_trips_from_csv(file_path)

    if trips:
        # 2.1. Сортировка по марке автомобиля
        print_trips_sorted_by_brand(trips)

        # 2.2. Сортировка по номеру
        print_trips_sorted_by_number(trips)

        # 2.3. Фильтрация по дате
        print_trips_filtered_by_date(trips)

        # 3. Добавление нового проезда
        add_trip_and_save(file_path, trips)

if __name__ == "__main__":
    main()
````

````python
import os
import csv
from datetime import datetime
from typing import List, Dict, Iterator

# Базовый класс для автомобиля
class Vehicle:
    def __init__(self, license_plate: str, car_brand: str):
        self.__setattr__('license_plate', license_plate)
        self.__setattr__('car_brand', car_brand)

    # Переопределение __setattr__ для записи значений в свойства
    def __setattr__(self, name: str, value) -> None:
        if name == 'license_plate' and not isinstance(value, str):
            raise ValueError("Номерной знак должен быть строкой")
        if name == 'car_brand' and not isinstance(value, str):
            raise ValueError("Марка автомобиля должна быть строкой")
        super().__setattr__(name, value)

    # Перегрузка __repr__
    def __repr__(self) -> str:
        return f"Vehicle(license_plate='{self.license_plate}', car_brand='{self.car_brand}')"

# Производный класс для описания проезда
class CarTrip(Vehicle):
    def __init__(self, number: int, date_time: datetime, license_plate: str, car_brand: str):
        super().__init__(license_plate, car_brand)
        self.__setattr__('number', number)
        self.__setattr__('date_time', date_time)

    # Переопределение __setattr__ с дополнительной валидацией
    def __setattr__(self, name: str, value) -> None:
        if name == 'number' and not isinstance(value, int):
            raise ValueError("Номер должен быть целым числом")
        if name == 'date_time' and not isinstance(value, datetime):
            raise ValueError("Дата и время должны быть объектом datetime")
        super().__setattr__(name, value)

    # Перегрузка __repr__
    def __repr__(self) -> str:
        return (f"CarTrip(number={self.number}, date_time='{self.date_time.strftime('%Y-%m-%d %H:%M:%S')}', "
                f"license_plate='{self.license_plate}', car_brand='{self.car_brand}')")

    # Статический метод для проверки формата даты
    @staticmethod
    def validate_date_format(date_str: str) -> bool:
        try:
            datetime.strptime(date_str, '%Y-%m-%d %H:%M:%S')
            return True
        except ValueError:
            return False

# Класс для управления коллекцией проездов
class TripCollection:
    def __init__(self, file_path: str):
        self.file_path = file_path
        self.trips: List[CarTrip] = []
        self.load_from_csv()

    # Реализация итератора
    def __iter__(self) -> Iterator[CarTrip]:
        return iter(self.trips)

    # Доступ к элементам по индексу
    def __getitem__(self, index: int) -> CarTrip:
        if 0 <= index < len(self.trips):
            return self.trips[index]
        raise IndexError("Индекс вне диапазона")

    # Генератор для фильтрации проездов по дате
    def trips_after_date(self, filter_date: datetime):
        for trip in self.trips:
            if trip.date_time > filter_date:
                yield trip

    # Загрузка данных из CSV
    def load_from_csv(self):
        try:
            with open(self.file_path, mode='r', encoding='utf-8') as file:
                reader = csv.DictReader(file)
                for row in reader:
                    number = int(row['№'])
                    date_time = datetime.strptime(row['дата и время'], '%Y-%m-%d %H:%M:%S')
                    license_plate = row['номерной знак']
                    car_brand = row['марка автомобиля']
                    trip = CarTrip(number, date_time, license_plate, car_brand)
                    self.trips.append(trip)
        except FileNotFoundError:
            print(f"Файл {self.file_path} не найден.")
        except Exception as e:
            print(f"Ошибка при чтении файла: {e}")

    # Сохранение данных в CSV
    def save_to_csv(self):
        try:
            with open(self.file_path, mode='w', encoding='utf-8', newline='') as file:
                fieldnames = ['№', 'дата и время', 'номерной знак', 'марка автомобиля']
                writer = csv.DictWriter(file, fieldnames=fieldnames)
                writer.writeheader()
                for trip in self.trips:
                    writer.writerow({
                        '№': trip.number,
                        'дата и время': trip.date_time.strftime('%Y-%m-%d %H:%M:%S'),
                        'номерной знак': trip.license_plate,
                        'марка автомобиля': trip.car_brand
                    })
            print("Данные успешно сохранены в файл.")
        except Exception as e:
            print(f"Ошибка при сохранении данных: {e}")

    # Сортировка по марке автомобиля
    def print_sorted_by_brand(self):
        if not self.trips:
            print("Нет данных для сортировки.")
            return
        sorted_trips = sorted(self.trips, key=lambda x: x.car_brand)
        print("\nПроезды, отсортированные по марке автомобиля:")
        for trip in sorted_trips:
            print(f"№: {trip.number}, Дата и время: {trip.date_time.strftime('%Y-%m-%d %H:%M:%S')}, "
                  f"Номерной знак: {trip.license_plate}, Марка: {trip.car_brand}")

    # Сортировка по номеру
    def print_sorted_by_number(self):
        if not self.trips:
            print("Нет данных для сортировки.")
            return
        sorted_trips = sorted(self.trips, key=lambda x: x.number)
        print("\nПроезды, отсортированные по номеру:")
        for trip in sorted_trips:
            print(f"№: {trip.number}, Дата и время: {trip.date_time.strftime('%Y-%m-%d %H:%M:%S')}, "
                  f"Номерной знак: {trip.license_plate}, Марка: {trip.car_brand}")

    # Фильтрация с использованием генератора
    def print_filtered_by_date(self, filter_date_str: str):
        try:
            filter_date = datetime.strptime(filter_date_str, '%Y-%m-%d %H:%M:%S')
            print(f"\nПроезды после {filter_date_str}:")
            for trip in self.trips_after_date(filter_date):
                print(f"№: {trip.number}, Дата и время: {trip.date_time.strftime('%Y-%m-%d %H:%M:%S')}, "
                      f"Номерной знак: {trip.license_plate}, Марка: {trip.car_brand}")
        except ValueError:
            print("Неверный формат даты. Используйте ГГГГ-ММ-ДД ЧЧ:ММ:СС")

    # Добавление нового проезда
    def add_trip(self):
        try:
            new_number = max(trip.number for trip in self.trips) + 1 if self.trips else 1
            print("\nДобавление нового проезда:")
            date_time_str = input("Введите дату и время (ГГГГ-ММ-ДД ЧЧ:ММ:СС): ")
            if not CarTrip.validate_date_format(date_time_str):
                raise ValueError("Неверный формат даты")
            date_time = datetime.strptime(date_time_str, '%Y-%m-%d %H:%M:%S')
            license_plate = input("Введите номерной знак: ")
            car_brand = input("Введите марку автомобиля: ")

            new_trip = CarTrip(new_number, date_time, license_plate, car_brand)
            self.trips.append(new_trip)
            self.save_to_csv()
        except Exception as e:
            print(f"Ошибка при добавлении проезда: {e}")

    # Статический метод для подсчёта файлов в директории
    @staticmethod
    def count_files_in_directory(directory_path: str) -> int:
        try:
            if not os.path.isdir(directory_path):
                raise FileNotFoundError(f"Директория {directory_path} не существует.")
            file_count = sum(1 for entry in os.scandir(directory_path) if entry.is_file())
            print(f"Количество файлов в директории '{directory_path}': {file_count}")
            return file_count
        except Exception as e:
            print(f"Ошибка при подсчёте файлов: {e}")
            return 0

# Основная программа
def main():
    # Создаём коллекцию проездов
    file_path = "data.csv"
    trips = TripCollection(file_path)

    # 1. Подсчёт файлов в директории (статический метод)
    directory_path = "."
    TripCollection.count_files_in_directory(directory_path)

    # Пример доступа по индексу
    if trips.trips:
        print("\nПример доступа по индексу (первый элемент):")
        print(trips[0])

    # Пример итерации по коллекции
    print("\nПример итерации по коллекции:")
    for trip in trips:
        print(trip)

    # 2.1. Сортировка по марке автомобиля
    trips.print_sorted_by_brand()

    # 2.2. Сортировка по номеру
    trips.print_sorted_by_number()

    # 2.3. Фильтрация по дате (с использованием генератора)
    trips.print_filtered_by_date("2023-10-15 00:00:00")

    # 3. Добавление нового проезда
    trips.add_trip()

if __name__ == "__main__":
    main()
````
